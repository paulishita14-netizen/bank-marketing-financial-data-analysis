## 6. Key Findings

- The dataset contains customers from different age groups, occupations, education levels, and marital-status categories.
- The majority of customers did not subscribe to the term deposit, while a smaller proportion subscribed.
- Customer age shows variation between subscribers and non-subscribers.
- Job and education categories show differences in term-deposit subscription rates.
- Housing-loan and personal-loan status are associated with differences in subscription rates.
- The method used to contact customers shows differences in subscription rates.
- The number of contacts during the current campaign varies considerably across customers.
- Previous campaign outcomes show noticeable differences in the likelihood of term-deposit subscription.
- Call duration differs between customers who subscribed and those who did not. However, call duration should be interpreted carefully because it is known after or during the customer interaction.
- Account balances vary considerably across customers, including both positive and negative balances.
- Contact month also shows differences in subscription rates.
- The analysis identifies associations between customer characteristics and subscription outcomes; these relationships should not be interpreted as proof of causation.

## 7. Conclusion

The Bank Marketing dataset provides useful information about customer characteristics, financial status, and campaign interactions. Through data cleaning and exploratory data analysis, different patterns in customer behavior and term-deposit subscription were examined.

The analysis considered variables such as age, job, education, marital status, account balance, housing loan, personal loan, contact method, campaign contacts, previous campaign outcome, call duration, and contact month. The results show that subscription rates vary across several of these customer and campaign characteristics.

The analysis also showed that the dataset is imbalanced, with a much larger number of customers who did not subscribe compared with those who subscribed. Therefore, percentages and subscription rates were used along with customer counts to provide a clearer understanding of the data.

Overall, the project demonstrates how financial customer data can be cleaned, explored, summarized, and visualized to identify meaningful patterns. These findings can support further analysis of customer behavior and can also provide a foundation for developing predictive models in future work.
